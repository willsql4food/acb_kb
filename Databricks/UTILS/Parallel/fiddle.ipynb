```python
import concurrent.futures
```

```python
def fn(i):
	return i**4
```

```python
with concurrent.futures.ThreadPoolExecutor() as tpExec:
    fns = set()
    fns.add(tpExec.submit(fn, 4))
    fns.add(tpExec.submit(fn, 8))
    
    print(type(fns), fns)

    for f in fns:
        print(f.result())
```
