```python
sql = """
WITH ranked_lineage AS (
  SELECT *,
    ROW_NUMBER() OVER (
      PARTITION BY target_table_full_name
      ORDER BY event_time DESC
    ) as rn
  FROM system.access.table_lineage
  WHERE entity_metadata.job_info.job_run_id IS NULL
    AND target_type IS NOT NULL
    and workspace_id = 1039620215775374
    --and source_table_catalog in ('steel_dev', 'silver_dev', 'edm_dev')
    --AND source_table_full_name = 'silver_prod.edm.dim_calendar'
)

--SELECT * EXCEPT(rn)
SELECT workspace_id, entity_type, entity_id, source_table_full_name, target_table_full_name, source_type, target_type
FROM ranked_lineage
WHERE rn = 1
"""

df = spark.sql(sql)

display(df)
```
