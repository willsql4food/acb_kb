# Function Parameters

A function parameter set can be thought of most simply as an _optional_ `tuple` followed by an _optional_ `dictionary`

* Positional - the `tuple`, conventially `*args`
	* Must be first in parameter list
	* Used based solely on where they show up
* Keyword - the `dictionary`, conventionally `**kwargs`
	* Name-value pairs of parameter arguments

BTW - `*` and `**` are _unpacking operators_ for `list` and `dictionary` respectively.

More info [here](https://docs.python.org/3/tutorial/controlflow.html#more-on-defining-functions)

```python
# A conventional function with space for positional and keyword parameters
def conventional_function(*args, **kwargs):
	# args is just a comma-separated set of values
	for a in args:
		print(a)

	# kwargs is a dictionary of key-value pairs
	for key, value in kwargs.items():
		print(f'{key} = {value}')


# But the names can be unconventional...
def nonstandard_function(*arrrgh, **matey):
	for a in arrrgh:
		print(a)

	for kw, val in matey.items():
		print(f'{kw} = {val}')
```

```python
# A tuple of tuples...
tpl = ('movie', 'Tenet'), ('star', 'Robert Pattinson')
# Used as a group of key-value pairs to build a dictionary
dct = dict(tpl)

# Another tuple to hold some random stuff...
args = 1, 99, 'hi', 'chicken', 4/3
# and a dictionary to hold some more random (but key-value paired) stuff
kwargs = {'pizza':'pepperoni', 'dumb':"Jim", 'dumber':"Bill", 'params':{"fish":"Nemo", "fowl":"Howard"}, 'dict_params':dct}

# Send the stuff we built above to the function, using unpacking operators to fill out the positional and keyword sections
conventional_function(*args, **kwargs)

print('=' * 80)

# If we don't unpack, it's just two complex positional arguments
conventional_function(args, kwargs)
```
