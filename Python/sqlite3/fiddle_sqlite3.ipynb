```python
import sqlite3
import csv
```

```python
def quotename(s: str):

    """Returns a string surrounded by quote character if the string contains spaces or tabs"""

    # Detect spaces or tabs in given string
    if s.find(' ') >= 0 or s.find('\t') >= 0:
        # Add escape character for existing double quotes
        s1 = s.replace('"', r'/"')
        # return the quoted string
        return f'"{s1}"'

    # otherwise, send back what was given
    else:
        return s

def test_quotename():
    assert quotename('hello') == 'hello'
    assert quotename('hello world') == '"hello world"'
    assert quotename('hello "world"') == '"hello /"world/""'
    assert quotename('') == ''
    assert quotename('hello\tworld') == '"hello\tworld"'

    print("all tests of quotename() pass")

test_quotename()
```
