# Sample of Logging with Python `logging` library

```python
import logging
```

## Setup
1. Set basic logging conditions - this determines lowest logging level for it and any subsequent handlers
1. Create a format for subsequent handlers
1. Create several additional handlers to catch different levels and send to different output
1. Clear any attached handlers (reruns add more!)
1. Add the defined handlers

```python
# Set up the default, basic logging handler:
#   DEBUG or higher messages written to a file
logging.basicConfig(level=logging.DEBUG,
					filename='log/onetime.log',
					filemode='w',
					format='%(asctime)s - %(levelname)s - %(message)s', 
					datefmt='%Y-%m-%d %H:%M:%S', 
					force=True)

# The output format desired for additional log handlers
format = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S')

# Setup a rotating file handler (keep five, each up to 1kb)
rotating_log = logging.handlers.RotatingFileHandler('log/logging_sample.log',
													maxBytes=1024, 
													backupCount=5)
rotating_log.setFormatter(format)
rotating_log.setLevel(logging.DEBUG)

# Setup a stream handler to put some messages to the console output
stream_log = logging.StreamHandler()
stream_log.setFormatter(format)
stream_log.setLevel(logging.WARNING)

# Get the logger for this instance, remove any existing handlers (in case of re-runs) and add the one defined above
logger = logging.getLogger(__name__)

for h in logger.handlers:
	logger.removeHandler(h)

logger.addHandler(rotating_log)
logger.addHandler(stream_log)
```

### Play!
Emit various messages and see where they landed

```python
logger.debug('This really is a debug message')
logger.info('FYI')
logger.warning('Alert!')
```
